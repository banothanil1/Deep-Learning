```python
from tensorflow.keras.preprocessing.text import one_hot

### below is the sentences
sent = [
    "the movie is good but the critics is not good",
    "the movie is excellent",
    "the love cup of tea",
    "iam a good boy",
    "iam a good developer",
    "i love watching coding tutorials and coding",
    "currently iam a php developer with better knowledge",
    "i work for ishan technologies which is a bad company"
]

vocab_size = 10000
dimension = 20
```

```python
onehot = [one_hot(words,vocab_size) for words in sent]
onehot
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_length = 15
padded_sequence = pad_sequences(onehot,padding='pre',maxlen=sent_length)
padded_sequence
```

```python
model = Sequential()
model.add(Embedding(vocab_size,dimension,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(padded_sequence[[0,1]])
```
